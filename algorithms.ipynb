```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def SGD(X: np.ndarray,
        Y: np.ndarray,
        Z: np.ndarray,
        lr: float = 0.1,
        steps: int = 50,
        seed: int = 42):
    """
    Simple Stochastic Gradient Descent on a surface.

    Parameters:
    - X, Y, Z: Meshgrid arrays of shape (N, N)
    - lr: Learning rate
    - steps: Number of optimization steps
    """

    np.random.seed(seed)
    trace_x, trace_y, trace_z = [], [], []

    # Pick a random starting index
    rows, cols = X.shape
    i = np.random.randint(0, rows)
    j = np.random.randint(0, cols)

    x, y = X[i, j], Y[i, j]

    for step in range(steps):
        # Compute gradient (numerical approximation)
        delta = 1e-3
        i_x = np.abs(X[0] - x).argmin()
        i_y = np.abs(Y[:,0] - y).argmin()
        
        z = Z[i_y, i_x]

        # Approximate gradient using finite differences
        z_dx = Z[i_y, np.clip(i_x + 1, 0, cols - 1)]
        z_dy = Z[np.clip(i_y + 1, 0, rows - 1), i_x]

        dz_dx = (z_dx - z) / delta
        dz_dy = (z_dy - z) / delta

        # Gradient descent update
        x = x - lr * dz_dx
        y = y - lr * dz_dy

        # Clamp to grid
        x = np.clip(x, X.min(), X.max())
        y = np.clip(y, Y.min(), Y.max())
        i_x = np.abs(X[0] - x).argmin()
        i_y = np.abs(Y[:,0] - y).argmin()
        z = Z[i_y, i_x]

        trace_x.append(x)
        trace_y.append(y)
        trace_z.append(z)

    return trace_x, trace_y, trace_z
```

```python
import numpy as np
import plotly.graph_objs as go


# SGD Trace (from earlier function)
tx, ty, tz = SGD(X, Y, Z, lr=0.2, steps=50)

# Surface
surface = go.Surface(z=Z, x=X, y=Y, colorscale='Viridis', opacity=0.7)

# Frames for animation
frames = []
for i in range(len(tx)):
    frame = go.Frame(
        data=[
            go.Scatter3d(
                x=tx[:i+1],
                y=ty[:i+1],
                z=tz[:i+1],
                mode='lines+markers',
                line=dict(color='red', width=5),
                marker=dict(size=5, color='black')
            )
        ],
        name=f'frame{i}'
    )
    frames.append(frame)

# Layout with rotating eye/camera
layout = go.Layout(
    title="SGD Path on 3D Surface",
    scene=dict(
        xaxis=dict(title='X'),
        yaxis=dict(title='Y'),
        zaxis=dict(title='Z'),
        camera=dict(
            eye=dict(x=1.25, y=1.25, z=1.25)  # starting eye position
        )
    ),
    updatemenus=[
        dict(
            type="buttons",
            buttons=[
                dict(label="Play",
                     method="animate",
                     args=[None, {
                         "frame": {"duration": 100, "redraw": True},
                         "fromcurrent": True,
                         "transition": {"duration": 0}
                     }]),
                dict(label="Pause",
                     method="animate",
                     args=[[None], {
                         "frame": {"duration": 0, "redraw": False},
                         "mode": "immediate",
                         "transition": {"duration": 0}
                     }])
            ]
        )
    ]
)

# Create figure
fig = go.Figure(
    data=[surface,
          go.Scatter3d(
              x=[tx[0]],
              y=[ty[0]],
              z=[tz[0]],
              mode='markers',
              marker=dict(size=5, color='red'))],
    layout=layout,
    frames=frames
)

fig.show()
```
